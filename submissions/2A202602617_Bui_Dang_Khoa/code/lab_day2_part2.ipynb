```python
import sys, numpy as np, torch
sys.path.insert(0, "/content/drive/MyDrive/deepweeds/code")
import train as TR
from eval import compute_metrics
from scipy.optimize import minimize_scalar

def nll(T, lg, y):
    p = TR.softmax(lg / T); return -np.log(p[np.arange(len(y)), y] + 1e-12).mean()

R = "/content/drive/MyDrive/deepweeds/runs/F01"
import pandas as pd
va = pd.read_csv("/content/data/val_subset0.csv")
for s in (0, 1, 2):
    lg = np.load(f"{R}/seed{s}/val_logits.npy")
    y = pd.read_csv(f"/content/drive/MyDrive/deepweeds/predictions/F01_seed{s}_val.csv").y_true.values
    T = minimize_scalar(lambda t: nll(t, lg, y), bounds=(0.3, 5), method="bounded").x
    p0, p1 = TR.softmax(lg), TR.softmax(lg / T)
    m0, m1 = compute_metrics(y, p0.argmax(1), p0), compute_metrics(y, p1.argmax(1), p1)
    print(f"seed{s}: T={T:.3f} | ECE val {m0['ece']:.4f} -> {m1['ece']:.4f} | NLL {m0['nll']:.4f} -> {m1['nll']:.4f} | F1 {m0['macro_f1']:.4f}={m1['macro_f1']:.4f}")
```

```python
import sys, time, numpy as np, pandas as pd, torch
sys.path.insert(0, "/content/drive/MyDrive/deepweeds/code")
import train as TR, dataset as D, model as M
from eval import compute_metrics
dev = torch.device("cuda")
cfg = TR.Config(exp_id="F01", seed=0)
va = pd.read_csv("/content/data/val_subset0.csv")
m = M.build_model("convnext_tiny", True, 9).to(dev).to(memory_format=torch.channels_last)
m.load_state_dict(torch.load(TR.run_dir(cfg)/"best.pt")); m.eval()

def logits(size, flip=False, amp=True):
    ld = D.make_loader(va, "/content/data", D.build_transforms(False, size), 128, False, None, 2, 0, True)
    out = []
    with torch.inference_mode():
        for x, y, _ in ld:
            x = x.to(dev)
            if flip: x = x.flip(3)
            with torch.autocast("cuda", dtype=torch.float16, enabled=amp):
                out.append(m(x).float().cpu().numpy())
    return np.concatenate(out), va.Label.values

def met(lg, y):
    p = TR.softmax(lg); r = compute_metrics(y, p.argmax(1), p)
    return round(r["macro_f1"],4), round(r["top1"],4), round(r["ece"],4)

def lat(size, K=1, amp=True, n=100):
    x = torch.randn(1,3,size,size, device=dev).contiguous(memory_format=torch.channels_last); t=[]
    with torch.inference_mode(), torch.autocast("cuda", dtype=torch.float16, enabled=amp):
        for _ in range(30): m(x)
        torch.cuda.synchronize()
        for _ in range(n):
            s = time.perf_counter()
            for _ in range(K): m(x)
            torch.cuda.synchronize(); t.append((time.perf_counter()-s)*1000)
    return [round(float(np.percentile(t,q)),1) for q in (50,95,99)]

l224, y = logits(224); lf, _ = logits(224, flip=True)
res = {"I00 1-view 224 (AMP)": (met(l224,y), lat(224)),
       "I01 TTA lật, gộp logit": (met((l224+lf)/2,y), lat(224,2)),
       "I03 TTA lật, gộp xác suất": (met(np.log((TR.softmax(l224)+TR.softmax(lf))/2),y), lat(224,2)),
       "I08 FP32 224": (met(logits(224,amp=False)[0],y), lat(224,amp=False))}
for s in (256, 288):
    res[f"I04 độ phân giải {s}"] = (met(logits(s)[0],y), lat(s))
for k, v in res.items(): print(k, "| F1,top1,ECE =", v[0], "| p50/p95/p99 ms =", v[1])
```

```python
import sys, numpy as np, pandas as pd, torch
sys.path.insert(0, "/content/drive/MyDrive/deepweeds/code")
import train as TR, dataset as D, model as M
from eval import compute_metrics
from scipy.optimize import minimize_scalar
dev = torch.device("cuda")
va = pd.read_csv("/content/data/val_subset0.csv"); y = va.Label.values
def val_logits(seed, size):
    cfg = TR.Config(exp_id="F01", seed=seed)
    m = M.build_model("convnext_tiny", True, 9).to(dev).to(memory_format=torch.channels_last)
    m.load_state_dict(torch.load(TR.run_dir(cfg)/"best.pt"))
    ld = D.make_loader(va, "/content/data", D.build_transforms(False, size), 128, False, None, 2, 0, True)
    return TR.evaluate(m, ld, torch.nn.CrossEntropyLoss(), dev, True)[2]
def met(p):
    r = compute_metrics(y, p.argmax(1), p); return round(r["macro_f1"],4), round(r["top1"],4), round(r["ece"],4)
nll = lambda T, lg: -np.log(TR.softmax(lg/T)[np.arange(len(y)), y]+1e-12).mean()
L288 = [val_logits(s, 288) for s in (0,1,2)]
T = minimize_scalar(lambda t: nll(t, L288[0]), bounds=(0.3,5), method="bounded").x
print("288 seed0 trước TS:", met(TR.softmax(L288[0])), "| T =", round(T,3), "| sau TS:", met(TR.softmax(L288[0]/T)))
L224 = [val_logits(s, 224) for s in (0,1,2)]
print("Ensemble 3 seed @224:", met(np.mean([TR.softmax(l) for l in L224], 0)))
print("Ensemble 3 seed @288:", met(np.mean([TR.softmax(l) for l in L288], 0)))
```

```python
%%writefile /content/drive/MyDrive/deepweeds/submission/2A202602617_Bui_Dang_Khoa/code/inference.py
from __future__ import annotations
import copy
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F


def _softmax(z):
    z = z - z.max(1, keepdims=True)
    e = np.exp(z)
    return e / e.sum(1, keepdims=True)


def predict_logits(model, loader, device, view=None, amp=True):
    """Logit theo đúng thứ tự file của loader. view: hàm biến đổi batch, hoặc None."""
    model.eval()
    fns, ys, lgs = [], [], []
    with torch.inference_mode():
        for x, y, f in loader:
            x = x.to(device)
            if view is not None:
                x = view(x)
            with torch.autocast("cuda", dtype=torch.float16, enabled=amp):
                lg = model(x)
            fns += list(f)
            ys.append(y.numpy())
            lgs.append(lg.float().cpu().numpy())
    return fns, np.concatenate(ys), np.concatenate(lgs)


def view_identity(x):
    return x


def view_hflip(x):
    return torch.flip(x, dims=[3])


def views_multicrop(x, crop: int):
    """5 crop (4 góc + giữa) và bản lật của chúng (10 view)."""
    H, W = x.shape[2], x.shape[3]
    pos = [(0, 0), (0, W - crop), (H - crop, 0), (H - crop, W - crop),
           ((H - crop) // 2, (W - crop) // 2)]
    crops = [x[:, :, i:i + crop, j:j + crop] for i, j in pos]
    return crops + [torch.flip(c, dims=[3]) for c in crops]


def views_multiscale(x, sizes):
    """Resize batch về từng kích thước. Chỉ dùng cho CNN/ConvNeXt (ViT/Swin cố định kích thước)."""
    return [F.interpolate(x, size=(s, s), mode="bicubic", align_corners=False) for s in sizes]


def aggregate_views(logits_per_view, space: str = "prob"):
    if space == "prob":
        return np.mean([_softmax(l) for l in logits_per_view], axis=0)
    if space == "logit":
        return _softmax(np.mean(logits_per_view, axis=0))
    raise ValueError(space)


def ensemble_probs(list_of_probs):
    return np.mean(list_of_probs, axis=0)


def fit_temperature(val_logits, val_labels) -> float:
    """Cực tiểu NLL trên VAL theo T (tìm có biên, scipy)."""
    from scipy.optimize import minimize_scalar
    y = np.asarray(val_labels)

    def nll(T):
        p = _softmax(val_logits / T)
        return -np.log(p[np.arange(len(y)), y] + 1e-12).mean()

    return float(minimize_scalar(nll, bounds=(0.3, 5), method="bounded").x)


def apply_temperature(logits, T: float):
    return _softmax(logits / T)


def fuse_conv_bn(model):
    """Gộp BN vào Conv liền trước (chỉ áp dụng cho Sequential/ResNet-style).
    ConvNeXt (LayerNorm), ViT, Swin không có BatchNorm: hàm trả về model không đổi."""
    model = copy.deepcopy(model).eval()

    def fuse(conv, bn):
        w = conv.weight
        b = conv.bias if conv.bias is not None else torch.zeros(w.size(0), device=w.device)
        s = bn.weight / torch.sqrt(bn.running_var + bn.eps)
        new = nn.Conv2d(conv.in_channels, conv.out_channels, conv.kernel_size, conv.stride,
                        conv.padding, conv.dilation, conv.groups, bias=True).to(w.device)
        new.weight.data = w * s.view(-1, 1, 1, 1)
        new.bias.data = bn.bias + s * (b - bn.running_mean)
        return new

    def walk(m):
        names = list(m._modules.keys())
        for i, n in enumerate(names):
            child = m._modules[n]
            walk(child)
            if isinstance(child, nn.BatchNorm2d) and i > 0 and isinstance(m._modules[names[i - 1]], nn.Conv2d):
                m._modules[names[i - 1]] = fuse(m._modules[names[i - 1]], child)
                m._modules[n] = nn.Identity()

    walk(model)
    return model
```

```python
%%writefile /content/drive/MyDrive/deepweeds/submission/2A202602617_Bui_Dang_Khoa/code/benchmark.py
from __future__ import annotations
import time
import numpy as np
import torch


def bench(fn, warmup: int = 10, iters: int = 100, sync=None) -> dict:
    sync = sync or (lambda: None)
    for _ in range(warmup):
        fn()
    sync()
    t = []
    for _ in range(iters):
        sync()
        t0 = time.perf_counter()
        fn()
        sync()
        t.append((time.perf_counter() - t0) * 1000)
    return {"p50": float(np.percentile(t, 50)), "p95": float(np.percentile(t, 95)),
            "p99": float(np.percentile(t, 99)), "mean": float(np.mean(t)), "n": iters}


def latency_report(model, batch_size: int, img_size: int, dtype: str = "fp32", device: str = "cuda",
                   warmup: int = 10, iters: int = 100) -> dict:
    model = model.to(device).eval()
    x = torch.randn(batch_size, 3, img_size, img_size, device=device)
    if dtype == "fp16":
        model = model.half(); x = x.half()
    sync = torch.cuda.synchronize if device == "cuda" else None

    def fn():
        with torch.inference_mode(), torch.autocast("cuda", dtype=torch.float16, enabled=(dtype == "amp")):
            model(x)

    r = bench(fn, warmup, iters, sync)
    return {"gpu": torch.cuda.get_device_name(0) if device == "cuda" else "cpu", "dtype": dtype,
            "batch": batch_size, "img_size": img_size, "p50": r["p50"], "p95": r["p95"], "p99": r["p99"],
            "images_per_s": batch_size / (r["p50"] / 1000), "torch": torch.__version__}


def tta_latency(model, k_views: int, **kw) -> dict:
    """Đo thật K lượt forward liên tiếp (so với K * p50 của 1 lượt)."""
    single = latency_report(model, **kw)
    model = model.eval()
    x = torch.randn(kw["batch_size"], 3, kw["img_size"], kw["img_size"], device=kw.get("device", "cuda"))

    def fn():
        with torch.inference_mode():
            for _ in range(k_views):
                model(x)

    r = bench(fn, kw.get("warmup", 10), kw.get("iters", 100),
              torch.cuda.synchronize if kw.get("device", "cuda") == "cuda" else None)
    return {"k": k_views, **{k: r[k] for k in ("p50", "p95", "p99")}, "single_p50": single["p50"],
            "k_times_single_p50": k_views * single["p50"]}
```
